Part 1: Tuples, lists that can't change

```python
point = (10, 20)
person = ("chinni", 28, "tenali")

print(person[0])      # chinni (indexing works like lists)
print(person[1:])     # (28, 'tenali') (slicing works)
print(len(person))    # 3

# person[0] = "ravi"  ← ERROR: tuples can't be changed

#  Trap: a single-item tuple needs a comma
a = (5)      # this is just the int 5
b = (5,)     # this is a tuple
print(type(a), type(b))   # <class 'int'> <class 'tuple'>
```

Unpacking (used everywhere in real code):

```python
name, age, city = person          # each variable gets one item
print(name, age, city)

# Swap two variables in one line (a favorite interview question)
x, y = 1, 2
x, y = y, x
print(x, y)    # 2 1

# * collects "the rest"
first, *rest = [1, 2, 3, 4]
print(first, rest)   # 1 [2, 3, 4]
nums = (1, 2, 2, 3)
print(nums.count(2))   # 2
print(nums.index(3))   # 3
```

Adding and removing:

```python
s = {"python", "sql"}
s.add("docker")            # add one item
print(s)
s.update(["aws", "git"])   # add many
print(s)
s.remove("sql")            # remove: ERROR if the item is missing
print(s)
s.discard("java")          # remove: NO error if missing (safer)
print(s)
item = s.pop()             # removes a random item
print(s, item)
```

Set math

```python
my_skills  = {"python", "sql", "git", "pandas"}
job_skills = {"python", "sql", "docker", "aws"}

print(my_skills & job_skills)   # intersection: skills in BOTH → {'python', 'sql'}
print(my_skills | job_skills)   # union: ALL skills combined
print(job_skills - my_skills)   # difference: what the job needs that I don't have → {'docker', 'aws'}
print(my_skills ^ job_skills)   # symmetric difference: in one but not both

# The same operations with method names
print(my_skills.intersection(job_skills))
print(job_skills.difference(my_skills))
```

Part 3: Dictionaries, the most important structure for real jobs

```python
person = {
    "name": "chinni",
    "age": 28,
    "city": "tenali",
    "skills": ["python", "sql"]
}

print(person["name"])        # chinni
print(person["skills"][0])   # python
print(len(person))           # 4 (number of keys)
```

Reading safely with [] vs .get():

```python
# print(person["salary"])           ← ERROR: KeyError (the key doesn't exist)
print(person.get("salary"))         # None (no error)
print(person.get("salary", 0))      # 0 (a default value you choose)
```

Looping (very important):

```python
scores = {"python": 90, "sql": 75, "git": 85}

print(scores.keys())     # dict_keys(['python', 'sql', 'git'])
print(scores.values())   # dict_values([90, 75, 85])
print(scores.items())    # pairs of (key, value)

for skill in scores:                 # loops over KEYS
    print(skill)

for skill, score in scores.items():  # loops over both (most used)
    print(f"{skill}: {score}")

print("python" in scores)            # True: `in` checks KEYS, not values
print(90 in scores.values())         # True
```

The counting pattern

```python
words = ["python", "sql", "python", "ai", "python", "sql"]
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1   # if new, start from 0, then add 1
print(counts)    # {'python': 3, 'sql': 2, 'ai': 1}
```

Part 4: Control flow, if / for / while 


if / elif / else

```python
score = 72

if score >= 90:
    print("Excellent")
elif score >= 70:
    print("Good")          # ← this runs
elif score >= 50:
    print("Pass")
else:
    print("Fail")

# Combine conditions
age = 28
has_degree = True
if age >= 21 and has_degree:
    print("Eligible")

# Truthiness (from Day 1): empty values count as False
skills = []
if not skills:
    print("No skills added yet")

# One-line if/else (conditional expression)
status = "pass" if score >= 50 else "fail"
print(status)
```

for loops and range()

```python
for i in range(5):          # 0 1 2 3 4 (stop is NOT included)
    print(i)

for i in range(1, 6):       # 1 to 5
    print(i)

for i in range(0, 11, 2):   # 0 2 4 6 8 10 (step 2)
    print(i)

for i in range(5, 0, -1):   # 5 4 3 2 1 (counting down)
    print(i)

for ch in "hi":             # loop over a string's characters
    print(ch)
```

while loops: repeat while a condition is true

```python
count = 1
while count <= 5:
    print(count)
    count += 1     # ⚠️ without this line, the loop runs forever
```

break and continue

```python
for n in range(1, 10):
    if n == 6:
        break          # exit the loop completely
    if n % 2 == 0:
        continue       # skip the rest of this round, go to the next number
    print(n)           # prints 1 3 5
```
