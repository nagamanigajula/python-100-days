DAY 3: Functions

```python
# Without a function: repeated code
print("=" * 20)
print("Welcome")
print("=" * 20)

# With a function: write once
def show_banner():
    print("=" * 20)
    print("Welcome")
    print("=" * 20)

show_banner()   # call it
show_banner()   # call it again: no repeated code
```

Part 2: Parameters and arguments

```python
def greet(name):              # name is a PARAMETER (a placeholder)
    print(f"Hello, {name}!")

greet("chinni")               # "chinni" is an ARGUMENT (the real value)
greet("ravi")

def introduce(name, age, city):
    print(f"{name}, {age}, from {city}")

introduce("chinni", 28, "tenali")    # positional: ORDER matters
# introduce("chinni", 28)            ← ERROR: missing 1 required argument 'city'
```

Part 3: return, getting a result back

```python
def add(a, b):
    return a + b          # send the result back to whoever called the function

result = add(3, 5)
print(result)             # 8
print(add(10, 20) * 2)    # 60: you can use the result directly
```

return vs print (the #1 beginner confusion):

```python
def add_print(a, b):
    print(a + b)          # only SHOWS the value on screen

def add_return(a, b):
    return a + b          # GIVES the value back so you can use it

x = add_print(2, 3)       # shows 5
print(x)                  # None ← nothing was returned!

y = add_return(2, 3)
print(y * 10)             # 50 ✅
```

```python
def check_age(age):
    if age < 0:
        return "Invalid age"     # stops here if the age is negative
    if age >= 18:
        return "Adult"
    return "Minor"

print(check_age(-5))   # Invalid age
print(check_age(28))   # Adult
```

```python
def get_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

low, high, avg = get_stats([4, 8, 1, 7])   # tuple unpacking
print(low, high, avg)                      # 1 8 5.0
```

Part 4: Default arguments

Rule: parameters with defaults must come after the ones without:

python
# def greet(greeting="Hello", name):   ← SyntaxError

```python
def greet(name, greeting="Hello"):     # greeting has a DEFAULT value
    return f"{greeting}, {name}!"

print(greet("chinni"))                 # Hello, chinni!
print(greet("chinni", "Namaste"))      # Namaste, chinni!
```

Famous interview trap: a mutable default argument

Why: the default list is created once, when the function is defined, and every call shares that same list.

```python
def add_skill(skill, skills=[]):       #  DANGEROUS
    skills.append(skill)
    return skills

print(add_skill("python"))   # ['python']
print(add_skill("sql"))      # ['python', 'sql'] ←  remembered from the last call!
```

correct way

```python
def add_skill(skill, skills=None):     # ✅
    if skills is None:
        skills = []                    # a fresh list on every call
    skills.append(skill)
    return skills

print(add_skill("python"))   # ['python']
print(add_skill("sql"))      # ['sql'] ✅
```

Part 5: Keyword arguments

```python
def create_profile(name, age, city):
    return f"{name} | {age} | {city}"

# Positional: order matters
print(create_profile("chinni", 28, "tenali"))

# Keyword: name each argument, so order doesn't matter
print(create_profile(city="tenali", name="chinni", age=28))

# Mixed: positional FIRST, then keyword
print(create_profile("chinni", city="tenali", age=28))
# Positional arguments must come before keyword arguments
print(create_profile("chinni", age=28, city="tenali"))
```
