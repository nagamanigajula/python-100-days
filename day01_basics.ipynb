Variables and data types

```python
# A variable is a name that stores a value
name = "Ravi"          # str  (text)
age = 24               # int  (whole number)
height = 5.8           # float (decimal number)
is_student = True      # bool (True or False)
salary = None          # NoneType (means "no value yet")

print(name, age, height, is_student, salary)

# type() tells you the data type
print(type(name))      # <class 'str'>
print(type(age))       # <class 'int'>
print(type(height))    # <class 'float'>
print(type(is_student))# <class 'bool'>
```

Number operations:

```python
a = 17
b = 5
print(a + b)    # 22  addition
print(a - b)    # 12  subtraction
print(a * b)    # 85  multiplication
print(a / b)    # 3.4 division (always gives float)
print(a // b)   # 3   floor division (drops the decimal)
print(a % b)    # 2   remainder (modulo), useful for even/odd checks
print(a ** 2)   # 289 power

# Shortcut operators
count = 10
count += 5      # same as count = count + 5  → 15
count -= 3      # → 12
count *= 2      # → 24
print(count)
```

Type conversion

```python
age_text = "24"
age_number = int(age_text)      # str → int
print(age_number + 1)           # 25

price = float("99.5")           # str → float
print(str(100) + " rupees")     # int → str to join text
print(int(7.9))                 # 7 (cuts the decimal, doesn't round)
print(round(7.9))               # 8
print(bool(0), bool(1), bool(""), bool("hi"))  # False True False True
```

Comparison and logic:

```python
x = 10
print(x > 5, x < 5, x == 10, x != 10, x >= 10)  # True False True False True
print(x > 5 and x < 20)   # True: both must be true
print(x > 50 or x == 10)  # True: at least one true
print(not x == 10)        # False: flips the result
```

Part 2: Strings, working with text

```python
text = "Python Developer"

# Indexing: positions start at 0
print(text[0])      # P
print(text[7])      # D
print(text[-1])     # r (negative counts from the end)

# Slicing: text[start:stop:step]; stop is NOT included
print(text[0:6])    # Python
print(text[7:])     # Developer (from 7 to end)
print(text[:6])     # Python (from start to 6)
print(text[::-1])   # repoleveD nohtyP (reversed)

print(len(text))    # 16 length
```

The most important string methods:

```python
s = "  Hello World  "
print(s.strip())          # "Hello World" removes spaces at both ends
print(s.lower())          # "  hello world  "
print(s.upper())          # "  HELLO WORLD  "
print(s.strip().title())  # "Hello World"

msg = "I love Java"
print(msg.replace("Java", "Python"))   # I love Python
print(msg.startswith("I"))             # True
print(msg.endswith("Java"))            # True
print(msg.find("love"))                # 2 (position; -1 if not found)
print(msg.count("a"))                  # 2
print("love" in msg)                   # True (membership check)

# split and join: used constantly
skills = "python,sql,docker"
skill_list = skills.split(",")         # ['python', 'sql', 'docker']
print(skill_list)
print(" | ".join(skill_list))          # python | sql | docker

# Check content
print("123".isdigit())    # True
print("abc".isalpha())    # True
```

```python
word = "hello"
# word[0] = "H"   ← this gives an ERROR
word = "H" + word[1:]   # make a new string instead
print(word)             # Hello
```

```python
name = "Ravi"
score = 92.456
print(f"Hi {name}, your score is {score}")
print(f"Rounded: {score:.2f}")       # 92.46 (2 decimals)
print(f"Next year age: {24 + 1}")    # you can put expressions inside {}
```

Part 3: Lists, the most-used data structure

```python
skills = ["python", "sql", "git"]
mixed = [1, "two", 3.0, True]     # lists can hold any types
empty = []

print(skills[0])      # python
print(skills[-1])     # git
print(skills[0:2])    # ['python', 'sql']
print(len(skills))    # 3
```

Adding items:

```python
skills.append("docker")           # add to END → ['python','sql','git','docker']
skills.insert(1, "fastapi")       # add at position 1
skills.extend(["aws", "rag"])     # add MANY items
print(skills)
```

Removing items:

```python
skills.remove("git")      # remove by VALUE (first match). Error if missing
last = skills.pop()       # remove and RETURN the last item
second = skills.pop(1)    # remove and return the item at position 1
del skills[0]             # delete by position
print(skills, last, second)
#skills.clear()  
#print(skills)        # empties the whole list
```

Changing, searching and counting:

```python
nums = [40, 10, 30, 10, 20]
nums[0] = 45              # change an item
print(nums.index(30))     # 2 (position of the first 30)
print(nums.count(10))     # 2 (how many 10s)
print(30 in nums)         # True
print(min(nums), max(nums), sum(nums))  # 10 45 115
```

Sorting:

```python
nums = [40, 10, 30, 20]
nums.sort()                   # changes the list itself → [10, 20, 30, 40]
nums.sort(reverse=True)       # → [40, 30, 20, 10]
nums.reverse()                # reverses the current order

original = [3, 1, 2]
new_sorted = sorted(original) # returns a NEW list; original unchanged
print(original, new_sorted)   # [3, 1, 2] [1, 2, 3]

words = ["banana", "kiwi", "apple"]
words.sort(key=len)   
print(words)        # sort by length → ['kiwi', 'apple', 'banana']
```

Looping through lists:

```python
skills = ["python", "sql", "git"]

for skill in skills:
    print(skill.upper())

for i, skill in enumerate(skills):   # gives the position AND the value
    print(i, skill)
```

```python
a = [1, 2, 3]
b = a            # NOT a copy! b and a point to the SAME list
b.append(4)
print(a)         # [1, 2, 3, 4]: a changed too!

c = a.copy()     # a real copy (a[:] also works)
c.append(99)
print(c)
print(a)         # [1, 2, 3, 4]: a is safe
```

```python
matrix = [[1, 2, 3], [4, 5, 6]]
print(matrix[1][2])   # 6 (row 1, column 2)
```
